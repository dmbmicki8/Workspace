### 1c. Are there any counties in which overdose deaths are trending downward?

```python
# From the data above, clearly, there is at least one county that trends downward.
query = """
    SELECT o.fipscounty, f.county, f.state, o.year, o.overdose_deaths
    FROM overdose_deaths o
    JOIN fips_county f ON o.fipscounty = CAST(f.fipscounty AS integer)
    ORDER BY o.fipscounty, o.year
"""

with engine.connect() as connection:
    all_counties_named = pd.read_sql(text(query), con=connection)

# Compare first year to last year
first_year = all_counties_named[all_counties_named['year'] == all_counties_named['year'].min()]
last_year = all_counties_named[all_counties_named['year'] == all_counties_named['year'].max()]

# Merge and compare
trend = first_year.merge(last_year, on=['fipscounty', 'county', 'state'], suffixes=('_first', '_last'))
trending_down_named = trend[trend['overdose_deaths_last'] < trend['overdose_deaths_first']]
trending_down_named[['county', 'state', 'overdose_deaths_first', 'overdose_deaths_last']]

# Most improved counties
trending_down_named = trend[trend['overdose_deaths_last'] < trend['overdose_deaths_first']].copy()
trending_down_named['improvement'] = trending_down_named['overdose_deaths_first'] - trending_down_named['overdose_deaths_last']
trending_down_named[['county', 'state', 'overdose_deaths_first', 'overdose_deaths_last', 'improvement']].sort_values('improvement', ascending=False)
```

### 2a. What is the correlation between spending on opioids and overdose deaths?

```python
# Will need to join the prescription, prescriber, and the cleaned drug table with the county data and overdose data.

query = """
    WITH drug_clean AS (
        SELECT DISTINCT drug_name, opioid_drug_flag
        FROM drug
    ),
    spending AS (
        SELECT pr.nppes_provider_zip5 AS zip,
               SUM(p.total_drug_cost) AS total_opioid_spending
        FROM prescriber pr
        JOIN prescription p ON p.npi = pr.npi
        JOIN drug_clean d ON p.drug_name = d.drug_name
        WHERE d.opioid_drug_flag = 'Y'
        GROUP BY pr.nppes_provider_zip5
    ),
    deaths AS (
        SELECT fipscounty, year, SUM(overdose_deaths) AS total_deaths
        FROM overdose_deaths
        GROUP BY fipscounty, year
    )
    SELECT d.year, d.fipscounty, d.total_deaths, SUM(s.total_opioid_spending) AS total_opioid_spending
    FROM deaths d
    JOIN zip_fips z ON CAST(z.fipscounty AS integer) = d.fipscounty
    JOIN spending s ON s.zip = z.zip
    GROUP BY d.year, d.fipscounty, d.total_deaths
    ORDER BY d.year
"""

with engine.connect() as connection:
    opioid_spending = pd.read_sql(text(query), con=connection)

opioid_spending.head(10)
```

```python
# Correlation between spending and overdoses:

correlation = opioid_spending['total_deaths'].corr(opioid_spending['total_opioid_spending'])
print(f'Correlation between opioid spending and overdose deaths: {correlation:.2f}')
```

```python
# Visualization of the correlation
plt.figure(figsize=(5,3))
plt.scatter(opioid_spending['total_opioid_spending'], opioid_spending['total_deaths'], alpha=0.5, color='red')
plt.title('Opioid Spending vs Overdose Deaths')
plt.xlabel('Total Opioid Spending ($)')
plt.ylabel('Total Deaths')
plt.grid(True)
plt.show()
```

##### counties that spend more on opioids tend to have significantly more overdose deaths.  As spending increases, deaths tend to increase as well.

### 2b. What is the ratio for spending on opioid vs non-opioid prescriptions?

```python
# Use CASE WHEN to find the ratio:

query = """
    WITH drug_clean AS (
        SELECT DISTINCT drug_name, opioid_drug_flag
        FROM drug
    )
    SELECT 
        SUM(CASE WHEN d.opioid_drug_flag = 'Y' THEN p.total_drug_cost ELSE 0 END) AS opioid_spending,
        SUM(CASE WHEN d.opioid_drug_flag = 'N' THEN p.total_drug_cost ELSE 0 END) AS non_opioid_spending
    FROM prescription p
    JOIN drug_clean d ON p.drug_name = d.drug_name
"""

with engine.connect() as connection:
    spending_ratio = pd.read_sql(text(query), con=connection)

spending_ratio['ratio'] = spending_ratio['opioid_spending'] / spending_ratio['non_opioid_spending']
print(f"Opioid to Non-Opioid spending ratio: {spending_ratio['ratio'][0]:.4f}")
spending_ratio
```

```python
# Visualization to show the breakdown of spending: 
plt.figure(figsize=(5,3))
plt.pie(
    [spending_ratio['opioid_spending'][0], spending_ratio['non_opioid_spending'][0]],
    labels=['Opioid', 'Non-Opioid'],
    colors=['green', 'lightblue'],
    autopct='%1.1f%%'
)
plt.title('Opioid vs Non-Opioid Spending')
plt.show()
```

### 2c. Are those who spend a higher ratio on opioids suffering from more deaths?

```python
query = """
    WITH drug_clean AS (
        SELECT DISTINCT drug_name, opioid_drug_flag
        FROM drug
    ),
    county_spending AS (
        SELECT 
            pr.nppes_provider_zip5 AS zip,
            SUM(CASE WHEN d.opioid_drug_flag = 'Y' THEN p.total_drug_cost ELSE 0 END) AS opioid_spending,
            SUM(CASE WHEN d.opioid_drug_flag = 'N' THEN p.total_drug_cost ELSE 0 END) AS non_opioid_spending
        FROM prescriber pr
        JOIN prescription p ON p.npi = pr.npi
        JOIN drug_clean d ON p.drug_name = d.drug_name
        GROUP BY pr.nppes_provider_zip5
    ),
    county_ratios AS (
        SELECT 
            z.fipscounty,
            SUM(cs.opioid_spending) AS opioid_spending,
            SUM(cs.non_opioid_spending) AS non_opioid_spending,
            SUM(cs.opioid_spending) / NULLIF(SUM(cs.non_opioid_spending), 0) AS opioid_ratio
        FROM county_spending cs
        JOIN zip_fips z ON z.zip = cs.zip
        GROUP BY z.fipscounty
    )
    SELECT cr.fipscounty, cr.opioid_ratio,
           SUM(o.overdose_deaths) AS total_deaths
    FROM county_ratios cr
    JOIN overdose_deaths o ON CAST(o.fipscounty AS varchar) = cr.fipscounty
    GROUP BY cr.fipscounty, cr.opioid_ratio
    ORDER BY cr.opioid_ratio DESC
"""

with engine.connect() as connection:
    ratio_vs_deaths = pd.read_sql(text(query), con=connection)

ratio_vs_deaths.head(10)
```

```python
correlation = ratio_vs_deaths['opioid_ratio'].corr(ratio_vs_deaths['total_deaths'])
print(f"Correlation between opioid ratio and overdose deaths: {correlation:.2f}")

plt.figure(figsize=(5,3))
plt.scatter(ratio_vs_deaths['opioid_ratio'], ratio_vs_deaths['total_deaths'], alpha=0.5, color='red')
plt.title('Opioid Spending Ratio vs Overdose Deaths')
plt.xlabel('Opioid Spending Ratio')
plt.ylabel('Total Deaths')
plt.grid(True)
plt.show()
```

##### Counties with a higher opioid spending ratio do not necessarily have more deaths. If anything, there's a very slight trend in the opposite direction, but it's so weak it's basically no relationship at all. To answer the overall question regarding the correlation of spending and OD deaths, there is almost no correlation. Counties that prioritized using opioids over other types of drugs are not necessarily seeing more deaths.

```python
### 3a.Which county has the highest overdose deaths per capita?
```

```python
#Multiply by 100k to get standard deaths

query = """
    SELECT f.county, f.state,
           SUM(o.overdose_deaths) AS total_deaths,
           AVG(p.population) AS avg_population,
           SUM(o.overdose_deaths) / AVG(p.population) * 100000 AS deaths_per_100k
    FROM overdose_deaths o
    JOIN fips_county f ON CAST(f.fipscounty AS integer) = o.fipscounty
    JOIN population p ON p.fipscounty = f.fipscounty
    GROUP BY f.county, f.state
    ORDER BY deaths_per_100k DESC
    LIMIT 10
"""

with engine.connect() as connection:
    deaths_per_capita = pd.read_sql(text(query), con=connection)

deaths_per_capita
```

```python
# Bar Chart visual to help see the results

plt.figure(figsize=(6, 4))
plt.barh(deaths_per_capita['county'], deaths_per_capita['deaths_per_100k'], color='orange')
plt.title('Overdose Deaths per 100k People\nTop 10 Counties')
plt.xlabel('Deaths per 100,000 People')
plt.gca().invert_yaxis()  # puts highest at top
plt.tight_layout()
plt.show()
```

### 3b.Which county has the most spending overall per capita?

```python
query = """
    WITH county_spending AS (
        SELECT 
            z.fipscounty,
            SUM(p.total_drug_cost) AS total_spending
        FROM prescription p
        JOIN prescriber pr ON pr.npi = p.npi
        JOIN zip_fips z ON z.zip = pr.nppes_provider_zip5
        GROUP BY z.fipscounty
    )
    SELECT f.county, f.state,
           cs.total_spending,
           AVG(pop.population) AS avg_population,
           cs.total_spending / AVG(pop.population) AS spending_per_capita
    FROM county_spending cs
    JOIN fips_county f ON f.fipscounty = cs.fipscounty
    JOIN population pop ON pop.fipscounty = f.fipscounty
    GROUP BY f.county, f.state, cs.total_spending
    ORDER BY spending_per_capita DESC
    LIMIT 10
"""

with engine.connect() as connection:
    spending_per_capita = pd.read_sql(text(query), con=connection)
    
spending_per_capita['total_spending'] = spending_per_capita['total_spending'].map('${:,.2f}'.format)
spending_per_capita['spending_per_capita'] = spending_per_capita['spending_per_capita'].map('${:,.2f}'.format)
spending_per_capita
```

### 3c. Which county has the most spending on opioids per capita?

```python
query = """
    WITH drug_clean AS (
        SELECT DISTINCT drug_name, opioid_drug_flag
        FROM drug
    ),
    county_opioid_spending AS (
        SELECT 
            z.fipscounty,
            SUM(p.total_drug_cost) AS total_opioid_spending
        FROM prescription p
        JOIN prescriber pr ON pr.npi = p.npi
        JOIN zip_fips z ON z.zip = pr.nppes_provider_zip5
        JOIN drug_clean d ON d.drug_name = p.drug_name
        WHERE d.opioid_drug_flag = 'Y'
        GROUP BY z.fipscounty
    )
    SELECT f.county, f.state,
           cs.total_opioid_spending,
           AVG(pop.population) AS avg_population,
           cs.total_opioid_spending / AVG(pop.population) AS opioid_spending_per_capita
    FROM county_opioid_spending cs
    JOIN fips_county f ON f.fipscounty = cs.fipscounty
    JOIN population pop ON pop.fipscounty = f.fipscounty
    GROUP BY f.county, f.state, cs.total_opioid_spending
    ORDER BY opioid_spending_per_capita DESC
    LIMIT 10
"""

with engine.connect() as connection:
    opioid_per_capita = pd.read_sql(text(query), con=connection)

opioid_per_capita['total_opioid_spending'] = opioid_per_capita['total_opioid_spending'].map('${:,.2f}'.format)
opioid_per_capita['opioid_spending_per_capita'] = opioid_per_capita['opioid_spending_per_capita'].map('${:,.2f}'.format)
opioid_per_capita
```

### 4a. Is there a correlation between unemployment rate and overdose deaths?

```python
# Use GROUP BY to AVG the monthly data into yearly figures to compare with OD deaths

query_unemp = """
    SELECT county, year, AVG(value) AS avg_unemployment
    FROM unemployment
    WHERE year BETWEEN 2015 AND 2018
    GROUP BY county, year
    ORDER BY county, year
"""

with engine_unemp.connect() as connection:
    unemp_by_county = pd.read_sql(text(query_unemp), con=connection)

unemp_by_county.head(10)
```

```python
# Merge the two databases to compare the data

# Pull overdose deaths with county names from prescribers database
query_deaths = """
    SELECT f.county, o.year, SUM(o.overdose_deaths) AS total_deaths
    FROM overdose_deaths o
    JOIN fips_county f ON CAST(f.fipscounty AS integer) = o.fipscounty
    GROUP BY f.county, o.year
    ORDER BY f.county, o.year
"""

with engine.connect() as connection:
    deaths_by_county = pd.read_sql(text(query_deaths), con=connection)

# Merge the two dataframes on county and year
unemp_vs_deaths = unemp_by_county.merge(deaths_by_county, on=['county', 'year'])

# Find correlation
correlation = unemp_vs_deaths['avg_unemployment'].corr(unemp_vs_deaths['total_deaths'])
print(f"Correlation between unemployment and overdose deaths: {correlation:.2f}")

unemp_vs_deaths.head(10)
```

### Where are the top 10 opioid prescribers located

```python
# ranks prescribers by total claims for opioid drugs
query = """
    WITH drug_clean AS (
        SELECT DISTINCT drug_name, opioid_drug_flag
        FROM drug
    )
    SELECT pr.npi,
           pr.nppes_provider_last_org_name AS last_name,
           pr.nppes_provider_first_name AS first_name,
           pr.nppes_provider_city AS city,
           pr.nppes_provider_state AS state,
           pr.nppes_provider_zip5 AS zip,
           SUM(p.total_claim_count) AS total_claims
    FROM prescriber pr
    JOIN prescription p ON p.npi = pr.npi
    JOIN drug_clean d ON d.drug_name = p.drug_name
    WHERE d.opioid_drug_flag = 'Y'
    GROUP BY pr.npi, pr.nppes_provider_last_org_name, 
             pr.nppes_provider_first_name, pr.nppes_provider_city,
             pr.nppes_provider_state, pr.nppes_provider_zip5
    ORDER BY total_claims DESC
    LIMIT 10
"""

with engine.connect() as connection:
    top_prescribers = pd.read_sql(text(query), con=connection)

top_prescribers
```

### 5b. Who is the top prescriber in each county? 

```python
# ranks prescribers within each county to show the #1 from each
query = """
    WITH drug_clean AS (
        SELECT DISTINCT drug_name, opioid_drug_flag
        FROM drug
    ),
    county_prescriber AS (
        SELECT pr.nppes_provider_last_org_name AS last_name,
               pr.nppes_provider_first_name AS first_name,
               pr.nppes_provider_city AS city,
               f.county,
               SUM(p.total_claim_count) AS total_claims,
               RANK() OVER (PARTITION BY f.county ORDER BY SUM(p.total_claim_count) DESC) AS rank
        FROM prescriber pr
        JOIN prescription p ON p.npi = pr.npi
        JOIN drug_clean d ON d.drug_name = p.drug_name
        JOIN zip_fips z ON z.zip = pr.nppes_provider_zip5
        JOIN fips_county f ON f.fipscounty = z.fipscounty
        WHERE d.opioid_drug_flag = 'Y'
        GROUP BY pr.nppes_provider_last_org_name, pr.nppes_provider_first_name,
                 pr.nppes_provider_city, f.county
    )
    SELECT county, last_name, first_name, city, total_claims
    FROM county_prescriber
    WHERE rank = 1
    ORDER BY county
"""

with engine.connect() as connection:
    top_per_county = pd.read_sql(text(query), con=connection)

top_per_county.head(10)
```

### What proportion of opioids are prescribed by the top 10 prescribers? Top 50? Top 100?

```python
query = """
    WITH drug_clean AS (
        SELECT DISTINCT drug_name, opioid_drug_flag
        FROM drug
    ),
    prescriber_totals AS (
        SELECT pr.npi,
               SUM(p.total_claim_count) AS total_claims
        FROM prescriber pr
        JOIN prescription p ON p.npi = pr.npi
        JOIN drug_clean d ON d.drug_name = p.drug_name
        WHERE d.opioid_drug_flag = 'Y'
        GROUP BY pr.npi
        ORDER BY total_claims DESC
    ),
    grand_total AS (
        SELECT SUM(total_claims) AS all_claims
        FROM prescriber_totals
    )
    SELECT 
        SUM(CASE WHEN rank <= 10 THEN total_claims ELSE 0 END) / MAX(all_claims) * 100 AS top_10_pct,
        SUM(CASE WHEN rank <= 50 THEN total_claims ELSE 0 END) / MAX(all_claims) * 100 AS top_50_pct,
        SUM(CASE WHEN rank <= 100 THEN total_claims ELSE 0 END) / MAX(all_claims) * 100 AS top_100_pct
    FROM (SELECT *, ROW_NUMBER() OVER (ORDER BY total_claims DESC) AS rank FROM prescriber_totals) ranked,
    grand_total
"""

with engine.connect() as connection:
    proportions = pd.read_sql(text(query), con=connection)

proportions
```

```python
plt.figure(figsize=(5,3))
plt.bar(['Top 10', 'Top 50', 'Top 100'], 
        [proportions['top_10_pct'][0], proportions['top_50_pct'][0], proportions['top_100_pct'][0]],
        color='red')
plt.title('% of Opioid Claims by Top Prescribers')
plt.xlabel('Prescriber Group')
plt.ylabel('% of Total Claims')
plt.show()
```
